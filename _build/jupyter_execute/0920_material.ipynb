# <font color='navy'>  Konštrukčný materiál </font>

Stavebnica *Stemfie-X* je navrhnutá pre použitie štandardného spojovacieho materiálu (skrutiek, matíc) bežne dostupného v železiarstvách alebo špecializovaných obchodoch. Pre záujmové krúžky je cenovo výtrazne výhodnejšie nakupovať celé balenia vo veľkobchodnej sieti, kde je zvyčajne dostupný aj širší sortiment materiálu.
